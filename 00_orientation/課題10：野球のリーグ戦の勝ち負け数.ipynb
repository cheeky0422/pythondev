```python
# Baseball_Team クラス定義
class Baseball_Team:
    def __init__(self, name: str, win: int, lose: int, draw: int):
        self.name = name
        self.win = win
        self.lose = lose
        self.draw = draw

    def calc_win_rate(self) -> float:
        """勝率 = 勝ち数 / (勝ち数 + 負け数)  ※引分は考慮しない"""
        games = self.win + self.lose
        return self.win / games if games > 0 else 0.0

    def show_team_result(self) -> None:
        """1行分の勝敗情報を表示"""
        print(f"{self.name:<9} {self.win:>3} {self.lose:>3} {self.draw:>3} {self.calc_win_rate():.3f}")

# 6チーム分のインスタンス生成
teams = [
    Baseball_Team("Archers",  77, 64, 2),
    Baseball_Team("Meteors",  71, 69, 3),
    Baseball_Team("Cheetahs", 69, 68, 6),
    Baseball_Team("Sharks",   70, 70, 3),
    Baseball_Team("Pythons",  68, 73, 2),
    Baseball_Team("Owls",     59, 82, 2),
]

# 見出し行を表示
print(f"{'Team':<9} {'W':>3} {'L':>3} {'D':>3} {'Win%':>5}")

# 各チームの結果を表示
for t in teams:
    t.show_team_result()
```
