```python
#@title 九九のリストを作成

# 九九の2次元リストを作成（リスト内包表記を利用）
kuku = [[v * h for h in range(1, 10)] for v in range(1, 10)]


#@title 九九の表を表示する
for row in kuku:                 # 1行ごとに取り出す
    for val in row:               # その行の値を取り出す
        print(f"{val:3d}", end="")  # 3桁幅で揃えて表示
    print("")                     # 行が終わったら改行
```
