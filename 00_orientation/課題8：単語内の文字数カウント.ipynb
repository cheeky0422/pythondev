```python
import string

# 1) 英単語を格納する空リスト
words = []

# 2) a〜z をキー、0 を値にした辞書
letter_counts = {ch: 0 for ch in string.ascii_lowercase}

# 3) 入力ループ（空行で終了）
while True:
    w = input("英単語を入力してください（未入力で終了）：").strip()
    if w == "":
        break

    words.append(w)  # 入力単語を都度リストに格納

    # 4) 文字ごとに出現回数を加算（a〜zのみ対象、大小区別なし）
    for ch in w.lower():
        if ch in letter_counts:      # 英字以外は無視
            letter_counts[ch] += 1

# 5) 結果
print("\n--- 入力結果 ---")
print("入力された英単語：", sorted(words, key=str.lower))  # ← リスト形式で表示

print("\n--- 文字の出現回数 ---")
for ch in string.ascii_lowercase:
    cnt = letter_counts[ch]
    if cnt > 0:
        print(f"{ch}が{cnt}個ありました")
```
